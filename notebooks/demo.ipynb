### Step 5: Visualize Processed Frame

```python
import cv2
import matplotlib.pyplot as plt

# Open output video and display middle frame
cap = cv2.VideoCapture("output/analyzed_traffic.mp4")
cap.set(cv2.CAP_PROP_POS_FRAMES, 50)
ret, frame = cap.read()
cap.release()

if ret:
    rgb_frame = cv2.cvtColor(frame, cv2.COLOR_BGR2RGB)
    plt.figure(figsize=(12, 7))
    plt.imshow(rgb_frame)
    plt.axis("off")
    plt.title("VisionGuard AI Processed Frame Telemetry", fontsize=14)
    plt.show()
```
